# 3.CCTV/WebCam Detection

## 3.1 Capture only one pic at a time

```python
front_face_detector = cv2.CascadeClassifier('haarcascade_frontalface_alt.xml')

video = cv2.VideoCapture(0) #0 --> web cam tqick 1--> anothe source cctv
success,frame = video.read() #Success is you capture inmage or not frame -->the exact image

if success == True:
    gray_image = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    face_detector = front_face_detector.detectMultiScale(gray_image)

    for x,y,w,h in face_detector:
        cv2.rectangle(gray_image,(x,y),(x+w,y+h),(255,0,0),3)
    cv2.imshow('image_show',gray_image)
    cv2.waitKey()

video.release()
```

## 3.2 You Press key capture image

```python
front_face_detector = cv2.CascadeClassifier('haarcascade_frontalface_alt.xml')

video = cv2.VideoCapture(0) #0 --> web cam tqick 1--> anothe source cctv

while True:
    success,frame = video.read()
    if success == True:
        gray_image = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        face_detector = front_face_detector.detectMultiScale(gray_image)
    
        for x,y,w,h in face_detector:
            cv2.rectangle(gray_image,(x,y),(x+w,y+h),(255,0,0),3)
        cv2.imshow('image_show',gray_image)
        cv2.waitKey()#Nothink in said pracet you don't go for next step wait key 1--> wait 1 millisecoun
       # nothing a i said this is not go next frame but 1 take another frame
       # give 1 move next frame for i millisecound
    
video.release()
```

## 3.3 Video live teacking gray image

```python
front_face_detector = cv2.CascadeClassifier('haarcascade_frontalface_alt.xml')

video = cv2.VideoCapture(0) #0 --> web cam tqick 1--> anothe source cctv

while True:
    success,frame = video.read()
    if success == True:
        gray_image = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        face_detector = front_face_detector.detectMultiScale(gray_image)
    
        for x,y,w,h in face_detector:
            cv2.rectangle(gray_image,(x,y),(x+w,y+h),(255,0,0),3)
        cv2.imshow('image_show',gray_image)
        key = cv2.waitKey(1)
        if key ==81 or key==113:#key vale use to break the loop
            break
        
video.release()
cv2.destroyAllWindows()
```

## 3.4 Video live tracking colored image

```python
front_face_detector = cv2.CascadeClassifier('haarcascade_frontalface_alt.xml')

video = cv2.VideoCapture(0) #0 --> web cam tqick 1--> anothe source cctv

while True:
    success,frame = video.read()
    if success == True:
        face_detector = front_face_detector.detectMultiScale(frame)
    
        for x,y,w,h in face_detector:
            cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),3)
        cv2.imshow('image_show',frame)
        key = cv2.waitKey(1)
        if key ==81 or key==113:#key vale use to break the loop
            break
        
video.release()
cv2.destroyAllWindows()
```
