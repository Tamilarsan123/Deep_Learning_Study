# Problem Statement
#### Build a machine learning model to accurately predict whether or not the patients in the dataset have diabetes or not?

### ======================================================================

# 1.Import Necessary libraries

```python
import pandas as pd
import keras
from sklearn.preprocessing import StandardScaler
```

# 2.Import Data

```python
diapedics_data = pd.read_csv('diabetes.csv')
diapedics_data
```

# 3.Data Understanding

```python
diapedics_data.shape
```

```python
diapedics_data.isna().sum()
```

```python
diapedics_data.dtypes
```

# 4.Data Preprocessing

```python
X = diapedics_data.drop('Outcome',axis=1)
y = diapedics_data['Outcome']
```

# 5.Model Building

```python
from keras.models import Sequential
from keras.layers import Dense,Dropout

import warnings
warnings.filterwarnings("ignore")
```

```python
model = Sequential()

model.add(Dense(12,input_dim=8,kernel_initializer='uniform',activation='relu'))
model.add(Dense(8,kernel_initializer='uniform',activation='relu'))

model.add(Dense(1,kernel_initializer='uniform',activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=46,stratify=y)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

### ALERT: Here's where You Need to apply STANDARDIZATION

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test  = scaler.transform(X_test)
```

# 6.Model Training || 7.Model Testing || 8.Model Evaluation

```python
model.fit(X_train,y_train,epochs=50,batch_size=16,validation_split=0.20)
```

# 9.Model Prediction 

```python
y_pred = model.predict(X_test)

y_pred = (y_pred > 0.60).astype(int)
y_pred
```

```python
model.save("diabetes_intelligence.h5")
```

### ======================================================================

## BONUS: Here you can check other Metrics for structured Data...

```python
from sklearn.metrics import confusion_matrix,classification_report

print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

### Write down the prediction pipeline you did while Training

##### input(array) -> standardization -> model -> result(will be in array) -> so use typecasting -> return the single output`

```python
from keras.models import load_model
intelligence_diabeties = load_model("diabetes_intelligence.h5")
```
