# LeNet 5 Architecture

## 1.Import Necessary Libraries

```python
import tensorflow as tf
from tensorflow.keras import layers,models
from tensorflow.keras.datasets import mnist
import numpy as np
```

## 2.load Dataset

```python
(X_train,y_train),(X_test,y_test) = mnist.load_data()
```

## 3. Data Preparation

#### 3.1 Data Normalaization

```python
X_train = X_train/255.0
X_test = X_test/255.0
```

#### 3.2 Reshape (Add chennel dimension)

```python
X_train = X_train.reshape(-1,28,28,1)# instead of we fill 60,000 images for batches, -1 says it to pick whatever we have fill it by yourself. In simple words, it is "auto-fill placeholder".
X_test = X_test.reshape(-1,28,28,1)
```

#### 3.3 Pad to 32x32(LeNet requirement)

```python
X_train = np.pad(X_train,((0,0),(2,2),(2,2),(0,0)),'constant')
X_test = np.pad(X_test,((0,0),(2,2),(2,2),(0,0)),'constant')
```

#### 3.4 One-Oht-Encoder

```python
y_train = tf.keras.utils.to_categorical(y_train,10)
y_test = tf.keras.utils.to_categorical(y_test,10)
```

## 4.Model Building

```python
model = models.Sequential()
```

```python
#C1 - Covalution
model.add(layers.Conv2D(6,kernel_size=(5, 5),activation='tanh',input_shape=(32, 32, 1)))

#S2 - Average Pooling 
model.add(layers.AveragePooling2D(pool_size=(2,2)))

#C3 - Convalution
model.add(layers.Conv2D(16,kernel_size=(5,5),activation='tanh'))

#S4 - Average Pooling
model.add(layers.AveragePooling2D(pool_size=(2,2)))

#C5 - Convalution(acts like fully connected)
model.add(layers.Conv2D(120,kernel_size=(5,5),activation='tanh'))

#Note:Remember: Convolution extracts features, pooling reduces spatial dimensions, and the output layer predicts which of the 10 digits the image represents.
```

```python
#Flatten 
model.add(layers.Flatten())

#F6 - Fully Connected 
model.add(layers.Dense(10,activation='tanh'))

#Output Layer
model.add(layers.Dense(10,activation='softmax'))
model.summary()
```

## 5.Moldel Complie

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

## 6.Model Training

```python
model.fit(X_train,y_train,epochs=10,batch_size=64,validation_data=(X_test,y_test))
```

## 7.Model Evaluation

```python
test_loss,test_acc = model.evaluate(X_test,y_test)
print("Test Accuracy:",test_acc)
```

## 8.Predict Example

```python
pred = model.predict(X_test[:5])
print("Predicted:", np.argmax(pred,axis=1))
print("Actual:",np.argmax(y_test[:5],axis=1))
```
